```python
# Customer Segmentation with K-Means + Visualizations
 # Internship Project - Data Science
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import confusion_matrix
import seaborn as sns
```

```python

# Step 1: Load dataset
from google.colab import files
uploaded = files.upload()   # upload Mall_Customers.csv
import pandas as pd
data = pd.read_csv("Mall_Customers.csv")
```

```python
# Step 2: Select features
X = data[['Annual Income (k$)', 'Spending Score (1-100)']]
```

```python
# Step 3: Standardize
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

```

```python
# Step 4: K-Means clustering
kmeans = KMeans(n_clusters=5, init='k-means++', random_state=42)
y_kmeans = kmeans.fit_predict(X_scaled)
data['Cluster'] = y_kmeans

```

```python
# Step 5: Scatter plot of clusters
plt.figure(figsize=(8,6))
sns.scatterplot(x=X_scaled[:,0], y=X_scaled[:,1], hue=y_kmeans, palette="Set2", s=100)
plt.title("Customer Segments")
plt.xlabel("Annual Income (scaled)")
plt.ylabel("Spending Score (scaled)")
plt.show()

```

```python
# Step 6: Bar chart - number of customers per cluster
plt.figure(figsize=(6,4))
data['Cluster'].value_counts().plot(kind='bar', color='skyblue')
plt.title("Number of Customers per Cluster")
plt.xlabel("Cluster")
plt.ylabel("Count")
plt.show()
```

```python
# Step 7: Pie chart - percentage of customers per cluster
plt.figure(figsize=(6,6))
data['Cluster'].value_counts().plot(kind='pie', autopct='%1.1f%%', colors=sns.color_palette("Set2"))
plt.title("Customer Distribution by Cluster")
plt.ylabel("")
plt.show()
```

```python
# Step 8: Visualize relationship between Gender and Cluster
# The previous code for confusion_matrix caused a ValueError due to mixed data types (string for Gender, int for Cluster).
# A confusion matrix is typically used for true vs. predicted labels in classification.
# For exploring the relationship between two categorical variables, pd.crosstab is more appropriate.
if 'Gender' in data.columns:
    # Create a crosstab (frequency table) of Gender against Cluster
    ct = pd.crosstab(data['Gender'], data['Cluster'])
    plt.figure(figsize=(6,4))
    sns.heatmap(ct, annot=True, fmt='d', cmap='Blues',
                xticklabels=ct.columns,
                yticklabels=ct.index)
    plt.title("Gender Distribution Across Clusters")
    plt.xlabel("Cluster")
    plt.ylabel("Gender")
    plt.show()
```

```python
# Step 9: Cluster summary
print(data.groupby('Cluster').mean(numeric_only=True))
```

```python
plt.savefig("scatter.png")
plt.savefig("heatmap.png")
plt.savefig("pei.png")
plt.savefig("box.png")
```
